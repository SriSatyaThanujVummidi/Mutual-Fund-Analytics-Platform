```python
nav_df    = pd.read_csv("data/raw/02_nav_history.csv")
master_df = pd.read_csv("data/raw/01_fund_master.csv")

# All AMFI codes in nav that exist in master
nav_codes    = set(nav_df['amfi_code'].unique())
master_codes = set(master_df['amfi_code'].unique())

missing_in_master = nav_codes - master_codes
missing_in_nav    = master_codes - nav_codes

print(f"✅ NAV codes      : {len(nav_codes)}")
print(f"✅ Master codes   : {len(master_codes)}")
print(f"⚠️  In NAV but NOT in master : {missing_in_master}")
print(f"⚠️  In master but NOT in NAV : {missing_in_nav}")

if not missing_in_master and not missing_in_nav:
    print("\n✅ All codes match perfectly!")
```
